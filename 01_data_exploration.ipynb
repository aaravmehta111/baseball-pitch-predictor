```python
from pybaseball import statcast_pitcher, playerid_lookup
player = playerid_lookup('webb','logan')
```

```python
import pandas as pd
```

```python
pitchers = [
    ('webb', 'logan'),
    ('lorenzen', 'michael'),
    ('yamamoto', 'yoshinobu'),
    ('gausman', 'kevin'),
    ('skenes', 'paul'),
    ('skubal', 'tarik'),
    ('chapman', 'aroldis'),
    ('doval', 'camilo'),
    ('crochet', 'garrett'),
    ('glasnow', 'tyler'),
]

all_data = []

for last_name, first_name in pitchers:
    lookup = playerid_lookup(last_name, first_name)
    pid = lookup['key_mlbam'].iloc[0]
    pitcher_data = statcast_pitcher('2025-04-01', '2025-10-31', pid)
    all_data.append(pitcher_data)

combined_data = pd.concat(all_data, ignore_index=True)
```
