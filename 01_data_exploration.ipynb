```python
import numpy as np

def random_guess_accuracy(y_test):
    classes = y_test.unique()
    np.random.seed(42)
    random_preds = np.random.choice(classes, size=len(y_test))
    return (random_preds == y_test.values).mean()

random_acc = random_guess_accuracy(y_test)
print(f"Random guess accuracy: {random_acc:.3f}")
```

```python
def most_common_baseline_accuracy(y_train, y_test):
    most_common = y_train.mode()[0]
    return (y_test == most_common).mean(), most_common

most_common_acc, most_common_class = most_common_baseline_accuracy(y_train, y_test)
print(f"Always-predict-'{most_common_class}' accuracy: {most_common_acc:.3f}")
```
